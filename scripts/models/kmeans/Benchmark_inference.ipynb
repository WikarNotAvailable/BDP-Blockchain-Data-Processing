```python
import boto3
from sagemaker import Session
from sagemaker.transformer import Transformer


region = "eu-north-1"
boto_session = boto3.Session(region_name=region)
sagemaker_client = boto3.client("sagemaker", region_name=region)
sagemaker_session = Session(boto_session=boto_session)

model_name = "kmeans-k-6"
output_path = f"s3://bdp-inference-results/kmeans/k_6/"

transformer = Transformer(
    model_name=model_name,
    instance_count=1,
    instance_type="ml.c5.2xlarge",
    strategy="MultiRecord",
    output_path=output_path,
    assemble_with="Line",
    accept="text/csv",
    sagemaker_session=sagemaker_session
)

try:
    transformer.transform(
        data='s3://bdp-test-data/scaled/test_data.csv',
        content_type="text/csv",
        split_type="Line",
        join_source="Input",
        input_filter="$[1:]",
        output_filter="$[0,-2,-1]"
    )
    print(f"Inference for k=6 completed. Results saved to: {output_path}")
except Exception as transform_error:
    print(f"Error during inference for k=6: {str(transform_error)}")
```
